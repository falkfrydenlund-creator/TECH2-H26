# Exercise 1: The Laffer curve

$$

\text{Revenue}(t) = t \times \text{Base}(t)
$$

strongly people respond to taxes.

    print the values $(t_{max}, \text{Revenue}(t_{max}))$.

    [`np.argmax()`](https://numpy.org/doc/stable/reference/generated/numpy.argmax.html)
    to find the position of the maximum.

    Augment the plot you created earlier with a vertical dashed line at
    $t_{max}$ and a black dot at the point $(t_{max}, \text{Revenue}(t_{max}))$.

    *Hint*: Use
    [`axvline()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.axvline.html)
    with `ls='--'` for the line, and
    [`scatter()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html)
    for the dot.
