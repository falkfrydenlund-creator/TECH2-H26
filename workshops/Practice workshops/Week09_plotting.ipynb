## Exercise 3: Choose a plot

Create a new column `REALRATE` for the real interest rate (`FEDFUNDS - INFLATION`).

1. Plot the real interest rate for every year.
2. Choose either a line or bar plot, and write one sentence explaining your choice.
3. Add a horizontal line at 0 with `plt.axhline()`. In which periods was the real interest rate negative?

**Check:** Can a reader understand each plot without looking at your code?

```python
annual['REALRATE'] = annual['FEDFUNDS'] - annual['INFLATION']
annual.head()
```

```python
fig, axes = plt.subplots(1,2, figsize=(12,4))

fig.suptitle('Line vs. bar plot')

axes[0].plot(annual['Year'],annual['REALRATE'])   
axes[0].axhline(color='black')
axes[0].set_title('Real Interest rate (line plot)')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Real interest rate(%)')

axes[1].bar(annual['Year'],annual['REALRATE'])
axes[1].axhline(color='black')
axes[1].set_title('Real Interest rate (bar plot)')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Real interest rate(%)')
```
